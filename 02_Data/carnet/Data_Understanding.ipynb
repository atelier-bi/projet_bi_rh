# L8 : Analyse des départs selon l'ancienneté

Créer des tranches d'ancienneté

```python

# Création des tranches d'ancienneté

df["Tranche_Anciennete"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, 20, 40],
    labels=["0-2 ans", "3-5 ans", "6-10 ans",
            "11-20 ans", "21 ans et plus"]
)

df["Tranche_Anciennete"].value_counts().sort_index()
```

Visualiser les effectifs par tranche

```python

# Répartition des salariés selon l'ancienneté

sns.countplot(data=df, x="Tranche_Anciennete")

plt.title("Répartition des salariés selon l'ancienneté")
plt.xlabel("Ancienneté")
plt.ylabel("Nombre de salariés")
plt.xticks(rotation=30)
plt.show()
```

Calculer les départs par tranche d'ancienneté

```python

# Analyse des départs selon l'ancienneté

anciennete = df.groupby(
    "Tranche_Anciennete", observed=True
)["Attrition"].agg(
    Effectif="count",
    Departs=lambda x: (x == "Yes").sum(),
    Proportion=lambda x: (x == "Yes").mean() * 100
)

anciennete["Proportion"] = anciennete["Proportion"].round(2)

anciennete
```

Visualiser la proportion de départs

```python

# Graphique des proportions de départs

sns.barplot(
    x=anciennete.index,
    y=anciennete["Proportion"]
)

plt.title("Proportion de départs selon l'ancienneté")
plt.xlabel("Ancienneté")
plt.ylabel("Proportion de départs (%)")
plt.xticks(rotation=30)
plt.show()
```


### Interprétation des départs selon l'ancienneté

Les salariés ayant entre 0 et 2 ans d'ancienneté
présentent la proportion de départs la plus élevée,
avec 29,82 %, soit 102 départs sur 342 salariés.

À l'inverse, les salariés ayant entre 11 et 20 ans
d'ancienneté présentent la proportion la plus faible
(6,67 %).

Ces résultats suggèrent que les salariés récemment
recrutés constituent un groupe prioritaire à examiner
pour les actions de fidélisation.


# L8 : Analyse des départs selon la satisfaction et les heures

Analyser les départs selon la satisfaction

```python

# Analyse des départs selon la satisfaction

satisfaction = df.groupby("JobSatisfaction")["Attrition"].agg(
    Effectif="count",
    Departs=lambda x: (x == "Yes").sum(),
    Proportion=lambda x: (x == "Yes").mean() * 100
)

satisfaction["Proportion"] = satisfaction["Proportion"].round(2)

satisfaction
```

Visualiser les résultats

```python

# Graphique des départs selon la satisfaction

sns.barplot(
    x=satisfaction.index,
    y=satisfaction["Proportion"]
)

plt.title("Départs selon la satisfaction au travail")
plt.xlabel("Niveau de satisfaction")
plt.ylabel("Proportion de départs (%)")
plt.show()
```

Analyser les départs selon les heures supplémentaires

```python

# Analyse des départs selon les heures supplémentaires

heures_sup = df.groupby("OverTime")["Attrition"].agg(
    Effectif="count",
    Departs=lambda x: (x == "Yes").sum(),
    Proportion=lambda x: (x == "Yes").mean() * 100
)

heures_sup["Proportion"] = heures_sup["Proportion"].round(2)

heures_sup
```

Visualiser les résultats

```python

# Graphique des départs selon les heures supplémentaires

sns.barplot(
    x=heures_sup.index,
    y=heures_sup["Proportion"]
)

plt.title("Départs selon les heures supplémentaires")
plt.xlabel("Heures supplémentaires")
plt.ylabel("Proportion de départs (%)")
plt.show()
```


### Conclusion de l'analyse exploratoire

L'analyse du dataset IBM HR Analytics a permis

# L9 : Diagnostic de la qualité des données




Cette étape consiste à vérifier la qualité du dataset
IBM HR Analytics.

Nous allons rechercher les valeurs manquantes,
les doublons et les valeurs aberrantes afin de
déterminer si les données sont fiables pour
notre analyse RH.


Vérifier les valeurs manquantes

```python

# Vérification des valeurs manquantes

valeurs_manquantes = df.isnull().sum()

print(valeurs_manquantes)
print("Total des valeurs manquantes :", valeurs_manquantes.sum())
```

Vérifier les doublons

```python

# Vérification des doublons

print("Lignes dupliquées :", df.duplicated().sum())

print("Identifiants dupliqués :",
      df["EmployeeNumber"].duplicated().sum())
```

Détecter les valeurs aberrantes

```python

# Détection des valeurs aberrantes avec IQR

colonnes = ["Age", "YearsAtCompany", "MonthlyIncome"]

for col in colonnes:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    borne_inf = Q1 - 1.5 * IQR
    borne_sup = Q3 + 1.5 * IQR

    aberrantes = df[
        (df[col] < borne_inf) |
        (df[col] > borne_sup)
    ]

    print(col, ":", len(aberrantes), "valeurs atypiques")
```


### Résultats du diagnostic qualité

L'analyse de la qualité des données a montré que :

- Aucune valeur manquante n'a été détectée.
- Aucune ligne dupliquée n'a été identifiée.
- La variable EmployeeNumber ne contient aucun doublon.
- La méthode IQR a détecté 104 valeurs atypiques
  pour YearsAtCompany et 114 pour MonthlyIncome.
- Aucune valeur atypique n'a été détectée pour Age.

Les valeurs atypiques ne seront pas supprimées
automatiquement, car elles peuvent correspondre
à des situations professionnelles réelles.

Le dataset présente donc une bonne qualité
structurelle. Des vérifications complémentaires
des règles métier restent nécessaires.
